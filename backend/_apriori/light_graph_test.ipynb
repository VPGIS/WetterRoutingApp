```python
def length_to_cm(G):
    """
    Wandelt Kantenlängen von Meter (float) in Zentimeter (int) um.
    """

    for _, _, _, data in G.edges(keys=True, data=True):
        if "length" in data:
            # Meter → cm und runden
            data["length"] = int(round(data["length"] * 100))

    return G
```

```python
import osmnx as ox

def graph_to_lightgraph(G, edge_attr_2keep=("length"), node_attr_2keep=()):
    """
    Generiert einen leichten Graphen für eine performante routen Berechnung.

    Parameter
    ----------
    G : networkx.MultiDiGraph
        OSMnx-Graph
    edge_attr_2keep: array
        Zu behaltende Edge Attribute für einen leichen Graphen.
    nodes_attr_2keep: array
        Zu behaltende Nodes Attribute für einen leichen Graphen.

    Returns
    -------
    G : networkx.MultiDiGraph
        leichter Graph ohne Geometrie:
    """

    # --- Edges bereinigen ---
    for _, _, _, data in G.edges(keys=True, data=True):
        for key in list(data.keys()):
            if key not in edge_attr_2keep:
                del data[key]

    # --- Nodes bereinigen ---
    for _, data in G.nodes(data=True):
        for key in list(data.keys()):
            if key not in node_attr_2keep:
                del data[key]

    return G
```
